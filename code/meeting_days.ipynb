# Generate Meeting Dates

```python
from datetime import date, timedelta
```

```python
FDOC = date(2026,6,22)
LDOC = date(2026,7,23)

FINAL = date(2026,7,28)

HOLIDAYS = {
    date(2026,7,3)
}
```

```python
def generate_class_days(start, end) -> list[date]:
    meetings: list[date] = []
    d = start
    while d <= end:
        if d.weekday() < 5 and not (d in HOLIDAYS): 
            # M - F : weekday is 0-4
            meetings.append(d)
        d = d + timedelta(days=1)

    return meetings
```

```python
meeting_days = generate_class_days(FDOC, LDOC)

print(len(meeting_days))
```

```python
def make_meeting_days(meeting_days):
    schedule: list[dict] = []
    for meeting in meeting_days:
        schedule.append({"date": meeting})
    return schedule


# So now, schedule[0] should be a dict representing the first day of class, and so on.
```
